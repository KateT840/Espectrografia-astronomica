```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Datos de las líneas de Fe I
# --------------------------------------------------

# Longitudes de onda en aire tomadas de tus datos de Moore (Å)
lambda_aire = np.array([
    4798.2649,
    4999.1125,
    5236.2041,
    5332.6602,
    5491.8315,
    5535.4179,
    5680.2404,
    6215.1438,
    6362.8763,
    6380.7433,
    7401.6849
])

# Anchos equivalentes de Moore (mÅ)
AE_mA = np.array([
    47, 29, 27, 45, 11, 113, 10, 40, 30, 40, 48
], dtype=float)

# Factores de oscilador f obtenidos de las filas de NIST
f = np.array([
    1.34e-2,
    3.9e-3,
    6.37e-3,
    1.7e-3,
    1.30e-3,
    1.39e-2,
    5.7e-4,
    9.57e-3,
    2.4e-3,
    8.41e-3,
    8.87e-3
])

# --------------------------------------------------
# 2. Convertir el ancho equivalente de mÅ a Å
# --------------------------------------------------

AE_A = AE_mA / 1000

# --------------------------------------------------
# 3. Probar un valor inicial de N
# --------------------------------------------------

# Este valor es SOLO una prueba inicial.
# Luego lo ajustaremos según la curva de crecimiento.
N = 1e14

# --------------------------------------------------
# 4. Calcular las coordenadas
# --------------------------------------------------

x = np.log10(N * f * lambda_aire / 5000)
y = np.log10(AE_A / lambda_aire)

# --------------------------------------------------
# 5. Mostrar la tabla con los valores calculados
# --------------------------------------------------

datos = pd.DataFrame({
    "Lambda (Å)": lambda_aire,
    "AE (mÅ)": AE_mA,
    "AE (Å)": AE_A,
    "f": f,
    "N": N,
    "x": x,
    "y": y
})

print(f"Valor provisional de N = {N:.2e}")
display(datos.round(6))

# --------------------------------------------------
# 6. Graficar los puntos
# --------------------------------------------------

plt.figure(figsize=(10, 6))
plt.scatter(x, y)

for i, lam in enumerate(lambda_aire):
    plt.annotate(
        f"{lam:.1f}",
        (x[i], y[i]),
        xytext=(4, 4),
        textcoords="offset points",
        fontsize=8
    )

plt.xlabel(r"$x = \log_{10}(N f \lambda / 5000\,\mathrm{\AA})$")
plt.ylabel(r"$y = \log_{10}(AE / \lambda)$")
plt.title(f"Curva de crecimiento preliminar para Fe I (N = {N:.1e})")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Valores de N para comparar
valores_N = [1e12, 1e13, 1e14, 1e15, 1e16]

plt.figure(figsize=(10, 6))

for N_prueba in valores_N:
    x_prueba = np.log10(N_prueba * f * lambda_aire / 5000)
    plt.scatter(x_prueba, y, label=f"N = {N_prueba:.0e}")

plt.xlabel(r"$x = \log_{10}(N f \lambda / 5000\,\mathrm{\AA})$")
plt.ylabel(r"$y = \log_{10}(AE / \lambda)$")
plt.title("Comparación de distintos valores de N")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('Comparacion_distintos_valores_N.png', dpi=300, bbox_inches='tight')
plt.show()

from google.colab import files
files.download('Comparacion_distintos_valores_N.png')
```
